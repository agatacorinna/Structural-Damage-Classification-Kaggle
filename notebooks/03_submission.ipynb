```python
import subprocess
import sys
from pathlib import Path
import pandas as pd
```

```python
root = Path.cwd()

if not(root / "scripts").exists():
    root = root.parent

result = subprocess.run(
    [sys.executable, "-m", "scripts.train"],
    cwd=root,
    capture_output=True,
    text=True
)
```

```python
print(result.stdout)
print(result.stderr)
print("EXIT CODE:", result.returncode)
```

```python

model_path = root / "models" / "final_model.pkl"

print("Model berhasil dibuat:", model_path.exists())
```

```python

result = subprocess.run(
    [sys.executable, "-m", "scripts.predict"],
    cwd=root,
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
print("EXIT CODE:", result.returncode)
```
